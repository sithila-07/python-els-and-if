## Odd number and Even number 

```python
number=int(input("Enter a number"))
if number%2==1:
    print("ODD Number")
else:
    print("Even Number")
```

## Greater Number

```python
number_1=int(input("Enter first number"))
number_2=int(input("Enter second number"))
if number_1==number_2:
    print("equal numbers")
else:
    if number_1>number_2:
        print("first number is greater than second number")
    else:
        print("second number is greater than first number")

```

## Checking three digit number

```python
number=int(input("Enter a number"))
if number>999:
    print("not a three digit number")
else:
    if number>99:
        print("three digit number")
    else:
        print("not a three digit number")
        
```

## grading system

```python
num=int(input("enter the marks"))
if num>=75:
    print("grade a")
elif num>=65:
    print("grade b")
elif num>=50:
    print("grade c")
elif num>=35:
    print("grade s")
else:
    print("fail")
    
```
